# Tile Sliding Domain: Initial State Space

```python
StateDimension = 3
InitialState = [1,2,3,4,5,6,0,7,8]
GoalState = [1,2,3,4,5,6,7,8,0]
Actions = lambda s: ['u', 'd', 'l', 'r']
OppositeActionsDict=dict([('u','d'),('d','u'),('l','r'),('r','l'), (None, None)])
```

```python
def Result(state: list[any], action: str) -> list[any]:
  '''state: initial state, action: u|d|l|r, returns new state'''
  i = state.index(0)
  newState = list(state)
  row,col=i//StateDimension, i % StateDimension
  if ( (action=='u' and row==0) or
       (action=='d' and row==StateDimension-1) or
       (action=='l' and col==0) or
       (action=='r' and col==StateDimension-1)):
      return newState
  if action=='u':
    l,r = row*StateDimension+col, (row-1)*StateDimension+col
  elif action=='d':
    l,r = row*StateDimension+col, (row+1)*StateDimension+col
  elif action=='l':
    l,r = row*StateDimension+col, row*StateDimension+col-1
  elif action=='r' :
    l,r = row*StateDimension+col, row*StateDimension+col+1
  newState[l], newState[r] = newState[r], newState[l]
  return newState

def PrintState(s: list[any]) -> None:
  for i in range(0,len(s),StateDimension):
    print(s[i:i+StateDimension])

def LegalMove(state: list[any], action: str) -> bool:
  '''shows if move was legal'''
  i = state.index(0)
  row,col=i//StateDimension, i % StateDimension
  # newState = state.copy()
  if ( (action=='u' and row==0) or
       (action=='d' and row==StateDimension-1) or
       (action=='l' and col==0) or
       (action=='r' and col==StateDimension-1)):
      return False
  return True
```

```python
def SingleTileManhattanDistance(tile: int, left: list[any], right: list[any]):
  ''' distance single tile'''
  leftIndex = left.index(tile)
  rightIndex = right.index(tile)
  return (abs(leftIndex//StateDimension-rightIndex//StateDimension) +
          abs(leftIndex%StateDimension-rightIndex%StateDimension))

def ManhattanDistance(left: list[any], right: list[any]):
  '''distance of two states'''
  distances = [SingleTileManhattanDistance(tile, left, right)
     for tile in range(1, StateDimension**2)]
  ### print ("Distances= ", distances)
  return sum(distances)
```

```python
def OutOfPlace(left: list[any], right: list[any]) -> int:
  '''shows how many tiles are out of place'''
  distances = [left[i]!=right[i] and right[i] != 0
     for i in range(StateDimension**2)]
  return sum(distances)
```

# Random Walk

Take some random moves from a state and return the new state and the sequence of moves.

Do not include moves undoing last move, or having no effect.

```python
def RandomWalk(state, steps: int) -> tuple[list[any], list[any]]:
  actionSequence = []
  actionLast = None
  for i in range(steps):
    action = None
    while action==None:
      action = random.choice(Actions(state))
      action = action if (LegalMove(state, action)
          and action!= OppositeActionsDict[actionLast]) else None

    actionLast = action
    state = Result(state, action)
    actionSequence.append(action)
  return state, actionSequence
```

```python
def ApplyMoves(actions: list[str], state: list[any]) -> list[any]:
  for action in actions:
    state = Result(state, action)
  return state
```

## Problem Class

INITIAL = InitialState  
IsGoal = Goal Test  
Actions = Actions List  
Result = Action Behavior  
ActionCost = Action Cost  

```python
class Problem(object):
    INITIAL = None
    IsGoal = None
    Actions = None
    Result = None
    ActionCost = None
```

## Node

```python
class Node(object):

  # initialization
  def __init__(self, state: list[any], parent = None, action: str = None, path_cost = 0):
    self.State=state
    self.Parent=parent
    self.Action=action
    self.PathCost = path_cost

  # as string
  def __str__(self) -> str:
    action = "<none>" if not self.Action else self.Action
    return str(self.State) + ", " + action

  # representation
  def __repr__(self) -> str:
    action = "<none>" if not self.Action else self.Action
    return str(self.State) + ", " + action

  # less than
  def __lt__(self, other) -> bool:
    return self.PathCost < other.PathCost;
```

## Expand

```python
def Expand(problem: Problem, node: Node) -> list[Node]:
  '''expands any possible node combinations'''
  ret = []
  s = node.State
  for action in problem.Actions(s):
    sPrime = problem.Result(s, action)
    cost = node.PathCost + problem.ActionCost(s,action,sPrime)
    ret.append(Node(sPrime, node, action, cost))
  return ret
```

## Breadth-First Search

```python
def BreadthFirstSearch(problem: Problem) -> tuple[Node, int]:
  node = Node(tuple(problem.INITIAL))
  if problem.IsGoal(node.State):
    return node, 0
  Frontier = []
  Frontier.append(node)
  reached = set()
  reached.add(tuple(problem.INITIAL))
  nodesExpanded = 0
  while (Frontier):
    ### print([str(n) for n in Frontier])
    node = Frontier.pop(0)
    ### print(node)
    for child in Expand(problem, node):
      s = tuple(child.State)
      ### print (s, "IsGoal=", problem.IsGoal(s))
      if problem.IsGoal(s):
        return child, nodesExpanded
      if s not in reached:
        reached.add(s)
        Frontier.append(child)
    nodesExpanded += 1
    if nodesExpanded > 500000:
      break;
  return None, nodesExpanded
```

## Best-First Search

```python
def BestFirstSearch(problem: Problem, f: Callable[[any], any]) -> tuple[Node, list[Node]]:
  node = Node(tuple(problem.INITIAL))
  Frontier = []
  heapq.heappush(Frontier,(f(node), node))
  reached = {}
  reached[tuple(problem.INITIAL)]=node
  nodesExpanded = 0
  while (Frontier):
    ##print([(x, str(n)) for (x,n) in Frontier])
    fValue, node = heapq.heappop(Frontier)
    ##print (node.State, "IsGoal=", problem.IsGoal(tuple(node.State)))
    if problem.IsGoal(tuple(node.State)):
      return node, nodesExpanded    ### print(node)
    for child in Expand(problem, node):
      s = tuple(child.State)
      if s not in reached or child.PathCost < reached[s].PathCost:
        reached[s] = child
        heapq.heappush(Frontier, (f(child), child))
    nodesExpanded += 1
    if nodesExpanded > 500000:
      break;
  return None, nodesExpanded
```

```python
def format_size(size_bytes):
    """Converts bytes to SI format"""
    if size_bytes == 0:
        return "0 B"
    units = ("B", "kB", "MB", "GB", "TB", "PB", "EB", "ZB", "YB")
    i = 0
    while size_bytes >= 1000 and i < len(units) - 1:
        size_bytes /= 1000
        i += 1
    return f"{size_bytes:.2f} {units[i]}"

def BenchmarkTileSliding(func: Callable[[Problem], tuple[Node, list[Node]]], pfunc: Callable[[int], Problem]) -> tuple[Node, list[Node]]:
    for steps in RandomWalkSteps:
        ITileSliding = pfunc(steps)
        StartTime = time.perf_counter()
        Ret, Cost = func(ITileSliding)
        EndTime = time.perf_counter()
        DeltaTime = (EndTime - StartTime) * 1000
        Sol = Solution(Ret)
        print("%02d" % (steps,), "steps", DeltaTime, "ms")
        print("\t           Initial:", ITileSliding.INITIAL)
        print("\t          Solution:", "".join(Sol))
        print("\t   Solution Length:", len(Sol))
        print("\t    Nodes Expanded:", Cost, " Size:", format_size(Cost * sys.getsizeof(Node)), "")

def AStarManhattan3x3(problem: Problem) -> tuple[Node, list[Node]]:
    AStarFb = lambda n: n.PathCost + ManhattanDistance(n.State, GoalState3x3)
    return BestFirstSearch(problem, AStarFb)
def AStarOutOfPlace3x3(problem: Problem) -> tuple[Node, list[Node]]:
    AStarFb = lambda n: n.PathCost + OutOfPlace(n.State, GoalState3x3)
    return BestFirstSearch(problem, AStarFb)
def BreadthFirst3x3(problem: Problem) -> tuple[Node, list[Node]]:
    return BreadthFirstSearch(problem)
def AStarManhattan4x4(problem: Problem) -> tuple[Node, list[Node]]:
    AStarFb = lambda n: n.PathCost + ManhattanDistance(n.State, GoalState4x4)
    return BestFirstSearch(problem, AStarFb)
def AStarOutOfPlace4x4(problem: Problem) -> tuple[Node, list[Node]]:
    AStarFb = lambda n: n.PathCost + OutOfPlace(n.State, GoalState4x4)
    return BestFirstSearch(problem, AStarFb)
def BreadthFirst4x4(problem: Problem) -> tuple[Node, list[Node]]:
    return BreadthFirstSearch(problem)
```

**Explain what these results indicate about search as a tool for Artificial Intelligence, especially as related to increase complexity in problem size.**

- Breadth-First Search (BFS)
 BFS struggles with large problem sizes due to its need to explore all possible states. This becomes especially costly in terms of time and memory. For instance, in the 4x4 Tile Sliding Puzzle, BFS generated a large number of nodes, particularly in scenarios involving long random walks.
- Uniform Cost Search (UCS)
 My experiments show that UCS performs similarly to BFS in terms of efficiency. While it does guarantee optimal solutions, its exhaustive approach leads to a steep rise in the number of nodes as the problem grows, making it resource-intensive.
- A Search with Manhattan Distance Heuristic*
 This version of A* shows clear improvements in efficiency. The Manhattan Distance heuristic helps direct the search toward the goal, significantly reducing the number of nodes expanded. It performed well on both 3x3 and 4x4 puzzles, producing solutions in acceptable time frames.
- A Search with Out-of-Place Heuristic*
 Although this heuristic improves performance over BFS and UCS, it's less effective than the Manhattan Distance heuristic. For both 3x3 and 4x4 puzzles, it resulted in more node expansions compared to A* with Manhattan Distance.
- Reflections on AI and Search Algorithms
All algorithms have their strong points and weak points, and operate differently based on the way that the problem is defined and its size. As problems grow bigger and more complicated such as a 3x3 vs. a 4x4 puzzle tile the effort and resources required by simple or uninformed search strategies may skyrocket. In such situations, generating and employing informed search strategies, particularly those that employ good heuristics, is quite valuable. The appropriate combination not only solves problems more rapidly but also consumes less space, enabling us to handle more complicated problems in the real world.
